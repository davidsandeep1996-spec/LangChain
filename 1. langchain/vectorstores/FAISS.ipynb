```python
from langchain_community.document_loaders import TextLoader
from langchain_community.document_loaders import CSVLoader
from langchain_community.embeddings import OllamaEmbeddings
from langchain_community.vectorstores import FAISS
from langchain.text_splitter import RecursiveCharacterTextSplitter


```

```python
# Load the CSV file
loader=CSVLoader(file_path='My_target_companies_jobs.csv', content_columns=['Company,Industry','What They Do','Leadership Team','Whitepapers/News','Pain Points','Agentic AI Opportunity','LinkedIn Contacts','Suggested Outreach','Status','Commented','page'])
documents=loader.load()
```

```python
# split the csv file into chunks
text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=512,
    chunk_overlap=100
)
texts = text_splitter.split_documents(documents)
```

```python
texts[0]  # first chunk of text
```

```python
# Create embeddings
embeddings=OllamaEmbeddings(model='deepseek-r1')
```

```python
# Create the vector store
db=FAISS.from_documents(texts,embeddings)
```

```python
# Retrieve similar documents
docs=db.similarity_search("what are the different Agentic ai opportunities?")
```

```python
# get the most similar document
docs[0].page_content  # content of the most similar document
```

```python
# db as retriever
ret=db.as_retriever()
docs=ret.invoke("what is the document is about?")
docs[0].page_content  # content of the most similar document
```
